## Step 3: Clean the Data🧹

 ### Fix inconsistent column names 🧼

```python
# Rename inconsistent columns if needed
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.head()
```

## Remove Duplicate Entries
### Check for duplicate rows and remove them if necessary.

```python
duplicated_data=df.duplicated()
```

```python
duplicated_data.sum()
```

```python
# Check for duplicates
df.duplicated().sum()
```

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
df
```

 ### Handle Missing Values ❓ 

```python
# Check for missing values
df.isnull().sum()
```

```python
df.isnull()
```

```python
# Fill or drop missing values
df['sleep_hours'] = df['sleep_hours'].fillna(df['sleep_hours'].mean())
```

```python
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].median())
```

```python
#new data
df.head()
```

```python
df
```

### Changing Wrong Data Types
Convert the data in the phone column to string instead of float

```python
df
```

```python
# Clean and convert phone column
df["phone"] = df["phone"].astype(str)  # Convert to string
```

```python
#cleaned data
df
```

### Standardize Values (e.g., Gender & Graduation_Eligible) 🔄 

```python
# Standardize 'gender' column
df['gender'] = df['gender'].replace({'F': 'Female', 'M': 'Male', 'Other' : 'Female'})
```

```python
#cleaned data
df
```

```python
# Graduation eligibility standardization
df['graduation_eligible'] = df['graduation_eligible'].replace({'Y': 'Yes', 'N': 'No', 'NO': 'No', 'no': 'No'})
```

```python
df
```

## Step 4: Explore the Data 🔍
### Which program has the highest average GPA? 🧠

```python
df.groupby('program')['gpa'].mean().sort_values(ascending=False)
```

### What’s the average sleep hours by level? 😴 

```python
df.groupby('level')['sleep_hours'].mean()
```

### Correlation Matrix 📈 

```python
df.corr(numeric_only=True)
```

## Step 5: Basic Visualization 🎨 

```python
#Importing necessary libraries
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# GPA Distribution
sns.histplot(df['gpa'], bins=20, kde=True)
plt.title('GPA Distribution')
plt.xlabel('GPA')
plt.ylabel('Frequency')
plt.show()
```

```python
# GPA by Program
sns.boxplot(x='program', y='gpa', data=df)
plt.xticks(rotation=45)
plt.title('GPA by Program')
plt.show()
```
